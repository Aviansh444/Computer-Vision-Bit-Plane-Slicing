```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

# Read grayscale image
img = cv2.imread("/content/drive/MyDrive/Computer Vision/pngtree-new-4k-nature-cb-editing-background-image_16123064.jpg", 0)

if img is None:
    print("Error: Image not found")
else:
    plt.figure(figsize=(12, 10))

    # Original Image
    plt.subplot(3, 3, 1)
    plt.imshow(img, cmap='gray')
    plt.title("Original Image")
    plt.axis('off')

    rows, cols = img.shape

    # Extract each bit plane manually
    for k in range(8):
        bit_plane = np.zeros((rows, cols), dtype=np.uint8)

        for i in range(rows):
            for j in range(cols):
                pixel = img[i, j]

                # Extract kth bit manually
                bit = (pixel // (2 ** k)) % 2

                if bit == 1:
                    bit_plane[i, j] = 255
                else:
                    bit_plane[i, j] = 0

        plt.subplot(3, 3, k + 2)
        plt.imshow(bit_plane, cmap='gray')
        plt.title(f"Bit Plane {k}")
        plt.axis('off')

    plt.tight_layout()
    plt.show()
```
